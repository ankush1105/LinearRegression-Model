```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('./Sample_dataset.csv')

print(data.head(3))

x = data.iloc[:,1:4]
y = data.SalePrice

print(x.head())
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.4, random_state= 42)

print(x_train.shape, y_train.shape, x_test.shape, y_test.shape)
```

```python
print(x_train.info)
print(x_test.info)
print(y_train.info)
print(y_test.info)
```

```python
#normalizing the LotArea column
print(x_train.head())
lota = np.array(x_train.LotArea)

lota = (lota - lota.min()) / (lota.max() - lota.min()) * 100

x_train.LotArea = lota

print(x_train.head())

lota_test = np.array(x_test.LotArea)
lota_test = (lota_test - lota_test.min()) / (lota_test.max() - lota_test.min()) * 100
x_test.LotArea = lota_test
print(x_test.head())
```

```python
class LinearRegression:
    def __init__(self, lr = 0.00001, epochs = 100):
        self.lr = lr
        self.epochs = epochs
        self.weights = None
        self.bias = 0

    def predict(self, xi):
        #print(f"weights {self.weights}\nxi {xi}\nmult : {self.weights * xi}")
        result = (self.weights * xi).sum() + self.bias
        return result

    def fit(self, params, targets, epochs = None):
        rows , features = params.shape
        self.weights = np.zeros(features)
        
        if epochs is None:
            epochs = self.epochs

        for i in range(epochs):
            #count_row = 0
            for xi, yi in zip(params.values, targets):
                #count_row += 1
                #print(f"i: {i} xi:{xi}, row: {count_row}/{rows}")
                predicted = self.predict(xi)
                #mse = (yi - predicted) ** 2
                #print(f"predicted: {predicted}, acutal: {yi}")
                gradient = 2  * (yi - predicted)

                self.weights += self.lr * gradient * xi

                self.bias += self.lr * gradient
            
            # print(f"weights: {self.weights}")
            # print(f"bias: {self.bias}")
```

```python
model = LinearRegression()

model.fit(x_train, y_train)
```

```python
predictions = []
errors = []

for xi, yi in zip(x_test.values, y_test):
    prediction = model.predict(xi)
    predictions.append(prediction)
    error = (prediction - yi)**2
    errors.append(error)

print(pd.Series(predictions, name= "Predictions"))

print(model.weights)

errors = np.array(errors)
print(errors.mean())
```
